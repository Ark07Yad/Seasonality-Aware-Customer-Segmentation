```python
import dask.dataframe as dd

print("Converting CSV to Parquet... (do this only once)")

df = dd.read_csv("combined_transactions.csv", low_memory=False)

# Save as parquet — compresses 42GB down to ~5-10GB
df.to_parquet(
    "transactions_parquet/",
    write_index=False,
    compression="snappy"       # fast compression, good balance
)

print("Done! You never need to do this again.")
```

```python
# Do the same for time.csv
df_time = dd.read_csv("time.csv", low_memory=False)
df_time.to_parquet("time_parquet/", write_index=False, compression="snappy")
print("time.csv converted!")
```
