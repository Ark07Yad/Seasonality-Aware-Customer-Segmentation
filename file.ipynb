```python
import pandas as pd
import glob
import os

# 1. Get all files starting with 'transactions_'
# If your script is NOT in the same folder, replace '.' with the full path
path = '.' 
all_files = sorted(glob.glob(os.path.join(path, "transactions_*.csv")))

if not all_files:
    print("❌ No files found! Check if the script is in the same folder as the data.")
else:
    output_file = "combined_transactions.csv"
    
    # 2. Process the first file to create the CSV and write the Header
    print(f"Starting with: {all_files[0]}")
    df_first = pd.read_csv(all_files[0])
    df_first.to_csv(output_file, index=False)
    
    # 3. Loop through the rest and APPEND (mode='a') without the header
    for filename in all_files[1:]:
        print(f"Appending: {filename}...")
        df = pd.read_csv(filename)
        # mode='a' means append; header=False prevents repeating column names
        df.to_csv(output_file, mode='a', index=False, header=False)

    print(f"✅ Success! All files combined into {output_file}")
```

```python
import pandas as pd

# 1. Load the time file
time_df = pd.read_csv('time.csv')
# Clean the time_df columns just in case
time_df.columns = time_df.columns.str.strip()

input_file = 'combined_transactions.csv'  # Your big combined file
output_file = 'final_dataset_with_dates.csv'

# 2. Check the columns of the big file first
# We read only the first row to save time/memory
sample = pd.read_csv(input_file, nrows=1)
print("Columns found in your transaction file:", sample.columns.tolist())

# Detect the correct 'shop_week' column name (ignoring case/spaces)
transaction_cols = {col.strip().lower(): col for col in sample.columns}
target_col = transaction_cols.get('shop_week')

if not target_col:
    print("❌ Error: Could not find a column related to 'shop_week' in your combined file.")
    print("Please check the list above and update 'on=...' in the merge code.")
else:
    print(f"✅ Found match: Using '{target_col}' to merge.")

    # 3. Process the big file in chunks
    first_chunk = True
    for chunk in pd.read_csv(input_file, chunksize=100000):
        # Merge using the detected column name
        merged_chunk = pd.merge(
            chunk, 
            time_df, 
            left_on=target_col, 
            right_on='shop_week', 
            how='left'
        )
        
        # Write to the new file
        if first_chunk:
            merged_chunk.to_csv(output_file, index=False, mode='w')
            first_chunk = False
        else:
            merged_chunk.to_csv(output_file, index=False, mode='a', header=False)

    print(f"Success! Saved to {output_file}")
```
